```python
def play_rock_paper_scissors(): 
    while True:
        player1 = input('Player 1, enter a choice: (rock, paper, scissors) ').lower().strip()
        player2 = input('Player 2, enter a choice: (rock, paper, scissors) ').lower().strip()

        choices = ["rock", "paper", "scissors"]

        if player1 not in choices or player2 not in choices:
            print("Invalid input detected. Please enter 'rock', 'paper', or 'scissors'.")
            continue 

        print(f"\nPlayer 1 chose {player1}, Player 2 chose {player2}.\n")

    # Determine the winner
        if player1 == player2:
            print("It's a draw!")
        elif player1 == 'rock':
            if player2 == 'scissors':
                print("Rock smashes scissors! Player 1 wins!")
            else:
                print("Paper covers rock! Player 2 wins!")
        elif player1 == 'paper':
            if player2 == 'rock':
                print("Paper covers rock! Player 1 wins!")
            else:
                print("Scissors cuts paper! Player 2 wins!")
        elif player1 == 'scissors':
            if player2 == 'paper':
                print("Scissors cuts paper! Player 1 wins!")
            else:
                print("Rock smashes scissors! Player 2 wins!")
        else:
            print("Invalid input detected. Please enter 'rock', 'paper', or 'scissors'.")

        play_again = input("\nDo you want to play again? (yes or no)?").lower().strip()
        if play_again != "yes":
            print("Thanks for playing!")
            break
       

while True:
    choice = input("Hello players! Do you want to play rock-paper-scissors? (yes or no)").lower().strip()
    if choice == 'yes':
        print("Great, let's play!")
        play_rock_paper_scissors()
        break
    elif choice == 'no':
        print("No worries, come back whenever you are ready.")
        break
    else: 
        print("Invalid input detected. Please enter yes or no.")



```
