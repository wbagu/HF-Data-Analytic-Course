```python
!mamba install pandas numpy matplotlib seaborn scikit-learn
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
```

```python
df = pd.read_csv('DOHMH_HIV_AIDS_Annual_Report_20260519.csv')
```

```python
#Data Cleaning: Removing all cammas on str and replacing the comma with a space/nothing. 
df = df.replace(',','', regex=True)
```

```python
# Data Cleaning: Converting the colums with data type objet to a numeric data type. Some were converted to int and others to float based on the type of numeric data found in each column. 
colsInt = ['HIV diagnoses', 'AIDS diagnoses', 'Deaths'] 
df[colsInt] = df[colsInt].apply(pd.to_numeric, errors = 'coerce').fillna(0).astype(int)

colsFlo = ['HIV diagnosis rate', '% linked to care within 3 months', 'AIDS diagnosis rate', 'PLWDHI prevalence', '% viral suppression', 'Death rate', 'HIV-related death rate', 'Non-HIV-related death rate'] 
df[colsFlo] = df[colsFlo].apply(pd.to_numeric, errors = 'coerce').fillna(0).astype(float)

df.head()
```

```python
# Data Cleaning: Replacing all NaN with 0 for better accessing/reading of the data. 
df = df.fillna(0).replace('NaN', 0)
```

```python
#Data Cleaning: Removing any row that contain the word All as this would be unusable for my analysis due to inconsistency on who data was entered as All. 
df = df[~df.astype(str).apply(lambda x: x.str.contains('All')).any(axis=1)]
df.head()
```

```python
#Data Statitis: Number of HIV diagnoses by year
HivSumByYear = df.groupby('Year')['HIV diagnoses'].sum()
display(HivSumByYear)
```

```python
#Data Statistic: Number of AIDS diagnoses by year
AidsSumByYear = df.groupby('Year')['AIDS diagnoses'].sum()
display(AidsSumByYear)
```

```python
#Data Statistic: Number of HIV diagnoses by Race by Borough
HivSumByRace = df.groupby(['Borough', 'Race'])['HIV diagnoses'].sum()
display(HivSumByRace)
```

```python
#Data Statistic: Number of HIV diagnoses by Age by Borough
HivSumByAge = df.groupby(['Borough', 'Age'])['HIV diagnoses'].sum()
display(HivSumByAge)
```

```python
#Data Statistic: Number of HIV Diagnoses by Borough, by Year, by Race: 
HivSumByAgeByRace = df.groupby(['Borough', 'Year', 'Race'])['HIV diagnoses'].sum()
pd.set_option('display.max_rows', None)
display(HivSumByAgeByRace)
```
