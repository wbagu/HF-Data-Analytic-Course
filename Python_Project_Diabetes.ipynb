```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
```

```python
df = pd.read_csv('https://raw.githubusercontent.com/niteen11/DataAnalyticsAcademy/master/Python/dataset_diabetes/diabetic_data.csv')
```

```python
plt.bar(df['race'] == 'Hispanic', x='gender', y='number_diagnoses')
```

```python
df.groupby(['race', 'gender'])['readmitted'].count()
```

```python
drop_cols = ['age', 'weight', 'admission_type_id', 'discharge_disposition_id', 'admission_source_id', 'time_in_hospital', 'payer_code', 'medical_specialty', 'num_lab_procedures', 'num_procedures', 'num_medications', 'number_outpatient', 'number_emergency', 'number_inpatient', 'diag_1', 'diag_2', 'diag_3', 'number_diagnoses', 'max_glu_serum', 'A1Cresult', 'metformin', 'repaglinide', 'nateglinide', 'chlorpropamide', 'glimepiride', 'acetohexamide', 'glipizide', 'glyburide', 'tolbutamide', 'pioglitazone', 'rosiglitazone', 'acarbose', 'miglitol', 'troglitazone', 'tolazamide', 'examide', 'citoglipton', 'insulin', 'glyburide-metformin', 'glipizide-metformin', 'glimepiride-pioglitazone', 'metformin-rosiglitazone', 'metformin-pioglitazone', 'change', 'readmitted']
new_df1 = df.drop(drop_cols, inplace=True, axis=1, errors='ignore')
```

```python
df.head()
```

```python
#Creating a new data frame but only with the data from my initial comparison point and measure.
dfh=df[df['race'] == 'Hispanic']
dfh.head()
```

```python
#Creating another data frame with the data from my comparison group.
dfa=df[df['race'] == 'Asian']
dfa.head()
```

```python
dfh.groupby(['gender', 'diabetesMed'])['diabetesMed'].count()
```

```python
dfh.info()
```

```python
#Creating a visual from my initial comparison group and measure.
categories = ['Female Yes', 'Female No', 'Male Yes', 'Male No']
Values = [826, 266, 725, 220]
plt.bar(categories, Values, color='skyblue', edgecolor='navy')
plt.title('Hispanic Patients - Diabetes Medication')
plt.show()
```

```python
dfa.groupby(['gender', 'diabetesMed'])['diabetesMed'].count()
```

```python
#Creating a second visual for my second comparison group.
categories = ['Female Yes', 'Female No', 'Male Yes', 'Male No']
Values = [240, 78, 235, 8]
plt.bar(categories, Values, color='pink', edgecolor='red')
plt.title('Asian Patients - Diabetes Medication')
plt.show()
```

#Comparing the public health data available we can conclude the Asian patients has a lower diabetic medication usage in comparison with the Hispanic patients. Doing the same comparison between both races by gender we can conclude there are more Asian female patients that do not take diabetes medication than Asian male patients which is a contract with the Hispanic patients were there is no big different within genders.
